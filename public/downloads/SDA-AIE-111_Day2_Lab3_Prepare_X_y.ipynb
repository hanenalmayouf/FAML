```python
import io
import pandas as pd
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]))
print(df.shape)
```

## 1) احفظ الهدف في y
`y` يحتوي الإجابة الصحيحة: 0 = مستمر، 1 = متوقف.

```python
y = df["churned_30d"].copy()
display(y.value_counts().sort_index())
```

## 2) أنشئ قائمة الأعمدة المستبعدة
اقرأ السبب بجانب كل مجموعة قبل تشغيل الكود.

```python
identifier = ["customer_id"]
target = ["churned_30d"]
future_information = [
    "refund_issued",
    "support_ticket_after_snapshot",
    "next_month_orders"
]
drop_cols = identifier + target + future_information
display(drop_cols)
```

## 3) أنشئ X
`drop()` ينشئ جدولًا جديدًا؛ لا يغير `df` الأصلي.

```python
X = df.drop(columns=drop_cols)
print("X:", X.shape)
print("y:", y.shape)
display(X.head(3))
```

## 4) اختبار التسرب
هذه الخلية يجب أن تنتهي برسالة نجاح.

```python
for forbidden in drop_cols:
    assert forbidden not in X.columns, f"يوجد عمود ممنوع داخل X: {forbidden}"
assert len(X) == len(y)
print("لا يوجد هدف أو معرّف أو معلومة مستقبل داخل X ✅")
```

## 5) صنّف الأعمدة قبل بناء Pipeline
شغّل الخلية، ثم راجع هل التصنيف منطقي.

```python
date_columns = ["signup_date", "snapshot_date"]
for column in date_columns:
    X[column] = pd.to_datetime(X[column])

numeric_columns = X.select_dtypes(include="number").columns.tolist()
categorical_columns = X.select_dtypes(include="object").columns.tolist()
datetime_columns = X.select_dtypes(include="datetime").columns.tolist()
print("عدد الأعمدة الرقمية:", len(numeric_columns))
print("عدد الأعمدة النصية/الفئوية:", len(categorical_columns))
print("عدد أعمدة التاريخ:", len(datetime_columns))
print("الفئوية:", categorical_columns)
print("التواريخ:", datetime_columns)
```
